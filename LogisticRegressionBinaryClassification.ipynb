```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
data={
    'age':[22,25,47,52,46,56,55,60,62,61,18,28,27,29,49],
    'have_insurance':[0,0,1,0,1,1,0,1,1,1,0,0,0,0,1]
}
df=pd.DataFrame(data)
print(df)
```

```python
plt.scatter(df.age,df.have_insurance,marker='+',color='red')
plt.xlabel('age')
plt.ylabel('have_insurance')
plt.show()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df[['age']],df.have_insurance,train_size=0.9)
```

```python
len(x_train)
len(x_test)
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
reg=linear_model.LogisticRegression()
reg.fit(x_train,y_train)
reg.predict(x_test)
```

```python
y_test
```

```python
reg.score(x_test,y_test)
```

```python
reg.predict_proba(x_test)
```
