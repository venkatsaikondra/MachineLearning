```python
import pandas as pd
import numpy as np
data={
    'name':['Braund,Owen','Cuming,john','Heikken,laina','Futrelle,Jacques','Allen,William','Moranm,james','Mccarthy,timotty','palsson,Gosta','johnson,oscar','Nasser,Nicholas'],
    'pclass':[3,1,3,1,3,3,1,3,3,2],
    'sex':['male','female','female','female','male','male','male','male','female','female'],
    'age':[22,38,26,35,35,np.nan,54,2,27,14],
    'sibsp':[1,1,0,1,0,0,0,3,0,1],
    'parch':[0,0,0,0,0,0,0,1,2,0],
    'ticket':[21171,17599,3101282,113803,373450,330877,17463,349909,347742,237736],
    'fare':[7.25,71.2833,7.925,53.1,8.05,8.4583,51.8625,21.075,11.1333,30.0708],
    'cabin':[np.nan,'c85',np.nan,'c123',np.nan,np.nan,'e46',np.nan,np.nan,np.nan],
    'embarked':['s','c','s','s','s','q','s','s','s','c'],
    'survived':[0,1,1,1,0,0,0,0,1,1]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
df.drop(['name','sibsp','parch','ticket','cabin','embarked'],axis='columns',inplace=True)
df
```

```python
target=df.survived
input=df.drop('survived',axis='columns')
```

```python
dummies=pd.get_dummies(input.sex).astype(int)
dummies
```

```python
inputs=pd.concat([input,dummies],axis='columns')
inputs
```

```python
inputs.drop('sex',axis='columns',inplace=True)
inputs
```

```python
inputs.age=inputs.age.fillna(inputs.age.mean())
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(inputs,target,test_size=0.3)
```

```python
len(x_train)
```

```python
from sklearn.naive_bayes import GaussianNB
model= GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```
