```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mayurdalvi/simple-linear-regression-placement-data")

print("Path to dataset files:", path)
```

```python
import pandas as pd
df = pd.read_csv('placement.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.cgpa,df.placed,marker='+',color='blue')
plt.xlabel('cgpa')
plt.ylabel('placed')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.placement_exam_marks,df.placed,marker='+',color='blue')
plt.xlabel('placement_exam_marks')
plt.ylabel('placed')
plt.show()
```

```python
from sklearn import linear_model
from sklearn.model_selection import train_test_split

# Create a logistic regression model
reg = linear_model.LogisticRegression()

# Split data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(
    df[['placement_exam_marks', 'cgpa']],
    df.placed,
    train_size=0.8                         #
)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
reg.fit(x_train,y_train)
```

```python
print("Training Accuracy:", reg.score(x_train, y_train))
print("Testing Accuracy:", reg.score(x_test, y_test))
```

```python
reg.predict([[80, 8.5]])
```
