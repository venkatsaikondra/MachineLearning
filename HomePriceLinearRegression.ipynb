```python
import pandas as pd
df=pd.read_excel('Rent.xlsx')
df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.scatter(df.area,df.rent,color='red',marker='o')
m,b=np.polyfit(df.area,df.rent,1)
plt.plot(df.area,m*df.area+b,color='blue')
plt.xlabel('area')
plt.ylabel('rent')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.area,df.rent,train_size=0.8)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn import linear_model
import numpy as np
x_train = np.array(x_train).reshape(-1, 1)
model=linear_model.LinearRegression()
model.fit(x_train,y_train)
```

```python
print("Training Accuracy:", model.score(x_train, y_train))
import numpy as np
x_test= np.array(x_test).reshape(-1, 1)
print("Testing Accuracy:", model.score(x_test, y_test))
```
