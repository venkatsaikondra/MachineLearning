```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
def gradient_decent(x,y):
    m_curr=b_curr=0
    iterations=1000
    n=len(x)
    learning_rate=0.001
    for i in range(iterations):
        y_predicted=m_curr*x+b_curr
        cost=(1/n)*sum([val**2 for val in (y-y_predicted)])
        md=-(2/n)*sum(x*(y-y_predicted))
        bd=-(2/n)*sum((y-y_predicted))
        m_curr=m_curr-learning_rate*md
        b_curr=b_curr-learning_rate*bd
        print("m {},b {}, cost{} iteration {}".format(m_curr,b_curr,cost,i) )
x=[1,2,3,4,5]
y=[5,7,9,11,13]
plt.plot(x,y,color='red',marker='o')
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,7,9,11,13])
gradient_decent(x,y)
```

```python
data={
    'name':['david','laura','sanjay','wei','jeff','aamir','venkat','virat','arthur','paul'],
    'math':[92,56,88,70,80,49,65,35,66,67],
    'cs':[98,68,81,80,83,52,66,30,68,73]
}
df=pd.DataFrame(data)
print(df)
```

```python
X = df[['math']]   # feature
y = df['cs']       # target

model = linear_model.LinearRegression()
model.fit(X, y)

# Predict values
y_pred = model.predict(X)

# Plot scatter + regression line
plt.scatter(df.math, df.cs, color='red', marker='o')
plt.plot(df.math, y_pred, color='blue')  # regression line
plt.xlabel('Math')
plt.ylabel('CS')
plt.title('Math vs CS with Regression Line')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Training data (area vs price)
X = np.array(df.math)
y = np.array(df.cs)

# Normalize X for stability
X = (X - X.mean()) / X.std()

# Gradient descent parameters
m = 0  # slope
b = 0  # intercept
alpha = 0.01  # learning rate
iterations = 1000
n = len(X)

# Store loss for plotting
loss_history = []

for _ in range(iterations):
    y_pred = m * X + b
    error = y_pred - y

    # Gradient descent update
    m -= alpha * (2/n) * np.dot(error, X)
    b -= alpha * (2/n) * np.sum(error)

    # Mean squared error
    mse = np.mean(error**2)
    loss_history.append(mse)

# Plot loss vs iterations
plt.plot(loss_history)
plt.xlabel("Iterations")
plt.ylabel("Mean Squared Error (MSE)")
plt.title("Gradient Descent Convergence")
plt.show()
```
